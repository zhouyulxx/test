```
ROUTE = {'purpose': 'candidate route inspection, not generation-ready approval', 'visual_route_verified': False, 'heading': 324, 'fov': 90, 'pitch': 0, 'size': [640, 360], 'rows': [{'index': 1, 'target_lat': 37.41160054638993, 'target_lng': -79.1393862038534, 'status': 'OK', 'pano_id': 'F2S9hlWmh9VW7syXf9P2gQ', 'lat': 37.41160054638993, 'lng': -79.1393862038534, 'date': '2023-11', 'offset_m': 0.0, 'gap_m': 0, 'bearing_deg': 324, 'distance_m': 0.0, 'image_http': 200, 'image_ok': True}, {'index': 2, 'target_lat': 37.41187948911811, 'target_lng': -79.13965684087397, 'status': 'OK', 'pano_id': 'ad7_c4bUo291BlAbiphZrg', 'lat': 37.41187948911811, 'lng': -79.13965684087397, 'date': '2023-11', 'offset_m': 0.0, 'gap_m': 39.2, 'bearing_deg': 322.4, 'distance_m': 39.2, 'image_http': 200, 'image_ok': True}, {'index': 3, 'target_lat': 37.41216824976726, 'target_lng': -79.1399143846171, 'status': 'OK', 'pano_id': 'MyCOLut5UrY04sfBDWNIcw', 'lat': 37.41216824976726, 'lng': -79.1399143846171, 'date': '2023-11', 'offset_m': 0.0, 'gap_m': 39.3, 'bearing_deg': 324.7, 'distance_m': 78.5, 'image_http': 200, 'image_ok': True}, {'index': 4, 'target_lat': 37.412422897800695, 'target_lng': -79.14014731521343, 'status': 'OK', 'pano_id': 'beOCA7I7o2mmDn0nppWGCw', 'lat': 37.4124126975654, 'lng': -79.14010121106251, 'date': '2025-09', 'offset_m': 4.2, 'gap_m': 31.8, 'bearing_deg': 328.7, 'distance_m': 110.3, 'image_http': 200, 'image_ok': True}, {'index': 5, 'target_lat': 37.41267754537714, 'target_lng': -79.14038024739348, 'status': 'OK', 'pano_id': 'BPlEq3jMKQo3z_mOuUk82Q', 'lat': 37.41265005708104, 'lng': -79.14034281871449, 'date': '2025-09', 'offset_m': 4.5, 'gap_m': 33.9, 'bearing_deg': 321.0, 'distance_m': 144.2, 'image_http': 200, 'image_ok': True}, {'index': 6, 'target_lat': 37.41293219249662, 'target_lng': -79.14061318115729, 'status': 'OK', 'pano_id': 'mDfJyCdnvIBZn3F9b5fcPw', 'lat': 37.41293616998295, 'lng': -79.14059441733069, 'date': '2025-09', 'offset_m': 1.7, 'gap_m': 38.8, 'bearing_deg': 325.1, 'distance_m': 183.1, 'image_http': 200, 'image_ok': True}, {'index': 7, 'target_lat': 37.41318683915911, 'target_lng': -79.14084611650486, 'status': 'OK', 'pano_id': 'WrfLBh7ama4reXQ40_OOhQ', 'lat': 37.41316674848713, 'lng': -79.14079933060322, 'date': '2025-09', 'offset_m': 4.7, 'gap_m': 31.4, 'bearing_deg': 324.8, 'distance_m': 214.4, 'image_http': 200, 'image_ok': True}, {'index': 8, 'target_lat': 37.4134414853646, 'target_lng': -79.14107905343624, 'status': 'OK', 'pano_id': 'Q_fokaeY89dhKTYhjjx94w', 'lat': 37.41348388025889, 'lng': -79.14108073032868, 'date': '2025-07', 'offset_m': 4.7, 'gap_m': 43.1, 'bearing_deg': 324.8, 'distance_m': 257.6, 'image_http': 200, 'image_ok': True}, {'index': 9, 'target_lat': 37.4136961311131, 'target_lng': -79.14131199195144, 'status': 'OK', 'pano_id': 't0tSGFt6n61Cz-kcNP8Odg', 'lat': 37.41371168927089, 'lng': -79.14128271400679, 'date': '2025-07', 'offset_m': 3.1, 'gap_m': 31.0, 'bearing_deg': 324.8, 'distance_m': 288.6, 'image_http': 200, 'image_ok': True}, {'index': 10, 'target_lat': 37.41395077640458, 'target_lng': -79.14154493205046, 'status': 'OK', 'pano_id': '7MForwrw4TZw_to6zDy-KA', 'lat': 37.41399511170184, 'lng': -79.14153835840416, 'date': '2025-09', 'offset_m': 5.0, 'gap_m': 38.8, 'bearing_deg': 324.4, 'distance_m': 327.3, 'image_http': 200, 'image_ok': True}, {'index': 11, 'target_lat': 37.414205421239046, 'target_lng': -79.14177787373337, 'status': 'OK', 'pano_id': 'q8ShrmMgZYu-272M9jrbwA', 'lat': 37.41421619237584, 'lng': -79.14173465631349, 'date': '2025-09', 'offset_m': 4.0, 'gap_m': 30.1, 'bearing_deg': 324.8, 'distance_m': 357.4, 'image_http': 200, 'image_ok': True}, {'index': 12, 'target_lat': 37.4144600656165, 'target_lng': -79.14201081700016, 'status': 'OK', 'pano_id': 'tpW66k9prWNMViv--AKjUA', 'lat': 37.41451749478453, 'lng': -79.14199994639287, 'date': '2025-09', 'offset_m': 6.5, 'gap_m': 40.9, 'bearing_deg': 325.0, 'distance_m': 398.3, 'image_http': 200, 'image_ok': True}]}
import json, hashlib, zipfile
from io import BytesIO
from datetime import datetime, timezone
import requests
from PIL import Image
from google.colab import userdata, files

def export_inputs():
    try:
        key = userdata.get("GOOGLE_MAPS_API_KEY").strip()
    except Exception:
        print("请启用 GOOGLE_MAPS_API_KEY 的 Notebook access。")
        return
    images = []
    anchors = []
    # Keep raw response bytes and attribution. No resizing or screenshot crops.
    for row in ROUTE["rows"]:
        label = f"P{row['index']:02d}"
        try:
            r = requests.get(
                "https://maps.googleapis.com/maps/api/streetview",
                params={"pano": row["pano_id"], "size": "640x360",
                        "heading": 324, "pitch": 0, "fov": 90,
                        "return_error_code": "true", "key": key}, timeout=30)
            if r.status_code != 200 or not r.headers.get("Content-Type", "").startswith("image/"):
                print(label, "下载失败，HTTP", r.status_code, "；未导出不完整包。")
                return
            im = Image.open(BytesIO(r.content))
            im.load()
            if im.size != (640,360) or im.format not in ("JPEG", "PNG"):
                print(label, "图片尺寸或格式异常；已停止。")
                return
            name = f"images/{label}." + ("jpg" if im.format == "JPEG" else "png")
            images.append((name, r.content))
            anchors.append({"label": label, "file": name,
                "sha256": hashlib.sha256(r.content).hexdigest(),
                "pano_id": row["pano_id"], "lat": row["lat"], "lng": row["lng"],
                "date": row["date"], "distance_m": row["distance_m"],
                "heading": 324, "pitch": 0, "fov": 90, "size": [640,360]})
            print(label, "OK")
        except Exception:
            print(label, "请求或图像校验失败；没有自动重试，未导出不完整包。")
            return
    manifest = {"schema": "streetview-local-anchors-v1", "source": "Google Street View Static API",
        "created_utc": datetime.now(timezone.utc).isoformat(),
        "route": "Lynchburg downtown straight route", "anchors": anchors,
        "note": "Mixed dates intentional. Coordinates do not calibrate model motion. Preserve attribution. Use and retention subject to provider agreement."}
    output = "lynchburg_12_inputs.zip"
    with zipfile.ZipFile(output, "w", compression=zipfile.ZIP_DEFLATED) as z:
        for name, content in images:
            z.writestr(name, content)
        z.writestr("anchors.json", json.dumps(manifest, indent=2))
    print("完成：12 张原始图片及 anchors.json；包中不含 Key。")
    files.download(output)

export_inputs()
```
